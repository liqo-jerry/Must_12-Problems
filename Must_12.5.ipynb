## Problem 5: Binary file with roll number, name and marks – Update marks

### 📌 Question

Create a binary file containing:

- Roll number
- Name
- Marks

Then enter a roll number and update the marks of that student.

### 🐍 Python Program

```python
import pickle

# Open binary file in write mode
file = open("student.dat", "wb")

# Enter number of students
n = int(input("Enter number of students: "))

# Enter student details
for i in range(n):

    roll = int(input("Enter roll number: "))
    name = input("Enter name: ")
    marks = float(input("Enter marks: "))

    # Create student record
    student = {
        "roll": roll,
        "name": name,
        "marks": marks
    }

    # Store the record in binary file
    pickle.dump(student, file)

# Close the file
file.close()


# Take roll number whose marks need to be updated
search_roll = int(input("Enter roll number to update: "))

# Take new marks
new_marks = float(input("Enter new marks: "))


# Read all records from the file
file = open("student.dat", "rb")

students = []

try:

    while True:

        # Read one record
        student = pickle.load(file)

        # Check roll number
        if student["roll"] == search_roll:

            # Update marks
            student["marks"] = new_marks

        # Add record to list
        students.append(student)

except EOFError:
    pass

# Close the file
file.close()


# Open the file again in write mode
file = open("student.dat", "wb")

# Write all records back into the file
for student in students:
    pickle.dump(student, file)

# Close the file
file.close()

print("Marks updated successfully.")
```
