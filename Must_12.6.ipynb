## Problem 6: Random number generator – Simulate a dice

### 📌 Question

Write a Python program that generates random numbers between 1 and 6, just like rolling a dice.

### 🐍 Python Program

```python
import random

# Generate a random number between 1 and 6
number = random.randint(1, 6)

# Display the result
print("Dice shows:", number)
```

## 📖 Explanation

First:

imports Python's random module.

Then:

generates a random integer between 1 and 6.

Possible outputs are:

- Dice shows: 1
- Dice shows: 2
- Dice shows: 3
- Dice shows: 4
- Dice shows: 5
- Dice shows: 6

Simulating multiple dice rolls

If we want to roll the dice 5 times:

```python
import random

for i in range(5):

    number = random.randint(1, 6)

    print("Dice shows:", number)
```
