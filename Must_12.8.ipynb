## Problem 8: Create a CSV file with User-ID and Password

### 📌 Question

Create a CSV file by entering:

- User-ID
- Password

Then read the CSV file and search for the password corresponding to a given User-ID.

### 🐍 Python Program

```python
import csv

# Open CSV file in write mode
file = open("users.csv", "w", newline="")

# Create CSV writer
writer = csv.writer(file)

# Write heading
writer.writerow(["User-ID", "Password"])

# Take number of users
n = int(input("Enter number of users: "))

# Enter user details
for i in range(n):

    user_id = input("Enter User-ID: ")
    password = input("Enter Password: ")

    # Write data into CSV file
    writer.writerow([user_id, password])

# Close the file
file.close()


# Search for a User-ID
search_id = input("Enter User-ID to search: ")

# Open CSV file in read mode
file = open("users.csv", "r")

# Create CSV reader
reader = csv.reader(file)

# Skip the heading
next(reader)

found = False

# Read each row
for row in reader:

    # Check User-ID
    if row[0] == search_id:

        print("Password =", row[1])
        found = True
        break

# Close the file
file.close()

# Display message if User-ID was not found
if found == False:

    print("User-ID not found.")
```
