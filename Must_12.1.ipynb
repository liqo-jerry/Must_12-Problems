### 📌 Question

## Problem 1: Read a text file line by line and display each word separated by # symbol.

- Example:

Suppose sample.txt contains:

Python is easy
I love programming
File handling is useful

- Output:

Python#is#easy
I#love#programming
File#handling#is#useful

### 🐍 Python Program

```python
# Open the text file in read mode
file = open("sample.txt", "r")

# Read the file line by line
for line in file:

    # Remove extra spaces and newline from the line
    line = line.strip()

    # Split the line into individual words
    words = line.split()

    # Display the words separated by #
    print("#".join(words))

# Close the file
file.close()
```

## 📖 Explanation

### Step 1: Open the file

file = open("sample.txt", "r")

- open() is used to open a file.
- "sample.txt" is the file name.
- "r" means read mode.

### Step 2: Read each line

for line in file:

- The for loop reads the file one line at a time.

### Step 3: Remove unnecessary characters

line = line.strip()

- strip() removes spaces and the newline character \n from the beginning and end of the line.

### Step 4: Separate words

words = line.split()

- For example:
Python is easy

- becomes:
["Python", "is", "easy"]

### Step 5: Join words using #

print("#".join(words))

- The join() function puts # between every word.
