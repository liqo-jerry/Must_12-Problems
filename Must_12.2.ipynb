### 📌 Question

## Problem 2: Count vowels, consonants, uppercase and lowercase characters

Read a text file and display the number of:

- Vowels
- Consonants
- Uppercase characters
- Lowercase characters

### 🐍 Python Program

```python
# Open the file in read mode
file = open("sample.txt", "r")

# Initialize counters
vowels = 0
consonants = 0
uppercase = 0
lowercase = 0

# Read the file character by character
for line in file:

    for ch in line:

        # Check whether the character is a vowel
        if ch in "aeiouAEIOU":
            vowels = vowels + 1

        # Check whether the character is an alphabet
        if ch.isalpha():

            # Check for consonant
            if ch not in "aeiouAEIOU":
                consonants = consonants + 1

        # Check whether the character is uppercase
        if ch.isupper():
            uppercase = uppercase + 1

        # Check whether the character is lowercase
        if ch.islower():
            lowercase = lowercase + 1

# Display the results
print("Number of vowels =", vowels)
print("Number of consonants =", consonants)
print("Number of uppercase characters =", uppercase)
print("Number of lowercase characters =", lowercase)

# Close the file
file.close()
```

## 📖 Explanation

We first create four counters:

- vowels = 0
- consonants = 0
- uppercase = 0
- lowercase = 0

#### Initially, all counters are 0.

- Checking vowels:

#### if ch in "aeiouAEIOU":

This checks whether the character is:
a e i o u
A E I O U

 If yes:

#### vowels = vowels + 1
The vowel counter increases by 1.

- Checking alphabet:

#### if ch.isalpha():

isalpha() returns True if the character is an alphabet.

- For example:

"A".isalpha()      # True
"5".isalpha()      # False

- Checking uppercase

#### if ch.isupper():

- For example:

"A".isupper()      # True
"a".isupper()      # False

- Checking lowercase

#### if ch.islower():

- For example:

"a".islower()      # True
"A".islower()      # False
