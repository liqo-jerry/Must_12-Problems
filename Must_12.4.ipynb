## Problem 4: Binary file with name and roll number – Search a roll number

### 📌 Question

Create a binary file containing:

- Roll number
- Name

Then search for a given roll number.

If found, display the student's name.
If not found, display an appropriate message.

For binary files, we will use Python's pickle module.

### 🐍 Python Program

```python
import pickle

# Open a binary file in write mode
file = open("student.dat", "wb")

# Take number of students
n = int(input("Enter number of students: "))

# Enter student details
for i in range(n):

    roll = int(input("Enter roll number: "))
    name = input("Enter name: ")

    # Store roll number and name in a dictionary
    student = {
        "roll": roll,
        "name": name
    }

    # Write the dictionary into the binary file
    pickle.dump(student, file)

# Close the file
file.close()

# Ask for roll number to search
search_roll = int(input("Enter roll number to search: "))

# Open the binary file in read mode
file = open("student.dat", "rb")

found = False

# Read records from the file
try:

    while True:

        # Read one student record
        student = pickle.load(file)

        # Compare roll number
        if student["roll"] == search_roll:

            print("Student Name =", student["name"])
            found = True
            break

except EOFError:
    # End of file reached
    pass

# Close the file
file.close()

# Display message if student was not found
if found == False:
    print("Roll number not found.")
```
