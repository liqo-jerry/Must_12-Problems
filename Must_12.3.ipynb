## Problem 3: Remove all lines containing the character 'a'


### 📌 Question

Read a text file and remove all lines that contain the character a.

The remaining lines should be written into another file.

### Example

## input.txt

Python is easy
Java is popular
Python programming
Data Science
Hello World
I love Python

#### If we remove lines containing a:

## output.txt

### 🐍 Python Program

```python
# Open the original file in read mode
input_file = open("input.txt", "r")

# Open another file in write mode
output_file = open("output.txt", "w")

# Read the file line by line
for line in input_file:

    # If 'a' is NOT present in the line
    if "a" not in line:

        # Write the line into the new file
        output_file.write(line)

# Close both files
input_file.close()
output_file.close()

print("Lines copied successfully.")
```

### Now let's check every line carefully:
