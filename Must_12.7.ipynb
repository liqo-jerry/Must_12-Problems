## Problem 7: Implement a Stack using List

### 📌 Question

Write a Python program to implement a stack using a list.

A stack follows:

- LIFO – Last In, First Out

Think of a stack of plates.

The plate placed last is removed first.

#### Stack Operations:

### 🐍 Python Program

```python
# Create an empty list for the stack
stack = []

while True:

    print("\n----- STACK MENU -----")
    print("1. Push")
    print("2. Pop")
    print("3. Peek")
    print("4. Display")
    print("5. Exit")

    choice = int(input("Enter your choice: "))

    # Push operation
    if choice == 1:

        item = input("Enter item: ")

        # Add item to the top of stack
        stack.append(item)

        print("Item pushed successfully.")

    # Pop operation
    elif choice == 2:

        # Check whether stack is empty
        if len(stack) == 0:

            print("Stack is empty.")

        else:

            # Remove the last item
            item = stack.pop()

            print("Removed item:", item)

    # Peek operation
    elif choice == 3:

        if len(stack) == 0:

            print("Stack is empty.")

        else:

            # Display the last item
            print("Top item:", stack[-1])

    # Display operation
    elif choice == 4:

        if len(stack) == 0:

            print("Stack is empty.")

        else:

            print("Stack:", stack)

    # Exit
    elif choice == 5:

        print("Program ended.")
        break

    # Invalid choice
    else:

        print("Invalid choice.")
```

### Important Stack Concepts

Push

adds an item to the stack.

- Example:

removes the last item.

[10, 20, 30, 40]

Pop

[10, 20, 30]

- 40 is removed because it was added last.

Peek

stack[-1]

- shows the last item without removing it.
